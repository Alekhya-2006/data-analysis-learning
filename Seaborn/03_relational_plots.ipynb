### Relational Plots
- Scatter Plots
- Line Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Scatter Plot
tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
)    
```

```python
# Line Plot

flights = sns.load_dataset("flights")
sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers"
)    
```
