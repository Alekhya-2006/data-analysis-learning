### Matrix Plots
- Heatmaps : Plot rectangular data as color-encoded matrix

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
flights = sns.load_dataset("flights")
flights.head()
```

```python
flights_pivot = flights.pivot(
    index = "month",
    columns = "year",
    values = "passengers"
)
flights_pivot
```

```python
sns.heatmap(
    flights_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
plt.title("Passengers Heatmap")
plt.tight_layout()
```
