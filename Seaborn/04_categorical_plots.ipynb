### Categorical plots
- Bar Plots
- Box Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
# Bar Plot
sns.barplot(
    data = tips,
    x = "day", 
    y = "tip",
    hue = "sex"
)
```

```python
# Box Plot
sns.boxplot(
    data = tips,
    x = "day", 
    y = "tip"
)
```
