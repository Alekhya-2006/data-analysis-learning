```python
import numpy as np
```

```python
# Mathematical functions

# aggregation
arr = np.array([1, 2, 3, 4, 5])

print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr)) # variance
print(np.argmin(arr)) # prints index of min 
print(np.argmax(arr)) # prints index of max
```

```python
# Power functions

print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr, 3))
```

```python
# log & exponent

print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))

print(np.exp(arr))
```

```python
# Rounding

print(np.round(3.14))
print(np.round(3.87))
print(np.ceil(3.14))
print(np.floor(3.14))
print(np.trunc(3.14))
```

```python
arr = np.array([1, -2, -3, 4, 2, -8, 4, 3, 5])

print(np.unique(arr))
print(np.sort(arr))
print(np.abs(arr))
```
