```python
import numpy as np
```

```python
# Normalizing an array means transforming its values so 
# they fit into a specific scale or range

arr = np.array([[1,2], [3, 4]])
mean = np.mean(arr) # mean
std_dev = np.std(arr) # standard deviation

normalized_arr = (arr - mean) / std_dev
print(normalized_arr)

print(np.mean(normalized_arr))  # mean = 0
print(np.std(normalized_arr)) # std deviation = 1
```
