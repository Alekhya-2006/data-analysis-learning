```python
import numpy as np
```

```python
# Data Types

arr1 = np.array([1, 2, 3, 4, 5])
print(arr1, arr1.dtype)

arr2 = np.array([1, 2, 3.14, 4, 5])
print(arr2, arr2.dtype)

arr3 = np.array([1, 2, 3, 4, "Tanuja"])
print(arr3, arr3.dtype)

arr4 = np.array([3+5j])
arr5 = np.array([2+3j])
print(arr4 + arr5, (arr4 + arr5).dtype)
print(arr4 - arr5, (arr4 - arr5).dtype)

arr = np.array(["prime", {1, 2, 3}, 3.14])
print(arr, arr.dtype)
```
