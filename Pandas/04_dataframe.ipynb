```python
import pandas as pd
```

```python
# DataFrame -> 2D labeled array

# 1. Dictionary of lists
info = {
    "Name": ["Anil Kumar", "Akhil", "Alekhya"],
    "Age": [25, 23, 20],
    "CGPA": [7.7, 9.8, 9]
}
data_frame = pd.DataFrame(info)
print(data_frame)
print(type(data_frame))
print(data_frame.index)
print(data_frame.columns)
```

```python
data_frame
```

```python
# 2. list of lists
df = pd.DataFrame([["Anil Kumar", 25], ["Akhil", 23], ["Alekhya", 20]], columns = ["Name", "Age"])
print(df)
```

```python
# 3. From numpy arrrays
import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])
df = pd.DataFrame(np_arr, columns = ["A", "B", "C"])
print(df)
```
