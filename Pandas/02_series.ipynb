```python
import pandas as pd
```

```python
# Series -> 1D labeled array

s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))
```

```python
# index -> position
# Index based access

print(s[0])
print(s[2])
```

```python
# label -> named index
s = pd.Series([20, 30, 40, 50], index = ['A', 'B', 'C', 'D'])
print(s)

# label based access
print(s['B'])
print(s['C'])

print(s[0]) # index based access
```
