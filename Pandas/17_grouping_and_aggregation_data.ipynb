```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")

df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()

df.groupby("country")["income"].agg(["min", "max","mean"])
df.groupby("country")["income"].aggregate(minimum_salary = "min", maximum_salary = "max", average_salary = "mean")

df.groupby("country").agg({
    "income": "mean",
    "age": "mean"
})

df.groupby("country").agg(
    avg_income = ("income", "mean"),
    avg_age = ("age", "mean")
)
```
