```python
import pandas as pd
```

```python
# Selecting Data

# select Columns
df = pd.read_csv("ipl_matches_2026.csv")
df[["city", "venue"]]
```

```python
# select rows (label and index based) - loc & iloc

# label
df.loc[1]
df.loc[0:2]  # [start idx: end idx] end index is inclusive here

# index based
df.iloc[0:2] # end idx exlusive here
```

```python
# select rows and columns

df.loc[0, "city"]
df.loc[0:2, ["team1", "team2"]]

# df.iloc[0, "city"] # error
df.iloc[0:2, 2] # 2 -> index of city in columns
```

```python
# select single scalar value - at(labels) & iat(idx)

df.at[0, "city"]
df.iat[0, 2]
```

```python
# copy vs view

cities = df["city"]
# cities[0] = "America" # it actually changes the original df 

# so when ever we want to create a copy 
cities = df["city"].copy()
cities[0] = "America"  # original is df is not changed
cities  
```
