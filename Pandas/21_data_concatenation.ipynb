```python
import pandas as pd
```

```python
# Data Concatenation
# Stacking DataFrames on top of each other(rows) or side by side(columns)

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Adam", "Eve", "Bob"]
})
df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["Charlie", "Fester", "Gilian"]
    
})

pd.concat([df1, df2]) # row
pd.concat([df1, df2], ignore_index = True)

pd.concat([df1, df2], axis = 1) # column
```
