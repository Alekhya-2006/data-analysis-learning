```python
import pandas as pd
```

```python
df = pd.read_csv("IRIS.csv")

df
```

```python
# A. Display the following info

# - first 10 rows
df.head(10)

# - shape and data types
df.shape
df.dtypes

# - summary Statistics( mean, std, min, max)
df.describe()
```

```python
# B. Select only those rows where
# - petal_length > 4.5
# - species = "Iris-virginica"

df[df["petal_length"] > 4.5]

df[df["species"] == "Iris-virginica"]

filtered_df = df[(df["petal_length"] > 4.5) & 
    (df["species"] == "Iris-virginica")]

print(filtered_df)
```

```python
# C. Group by species and compute
# - Average sepal_length
# - Maximum petal_width
# - Standard deviation of sepal_width

df.groupby("species").agg(
    avg_sepal_length=("sepal_length", "mean"),
    max_petal_width=("petal_width", "max"),
    std_sepal_width=("sepal_width", "std")
)
```

```python
# D. Create a new column : "petal_ratio" = petal_length / petal_width
# - For each species, find the average petal_ratio

df["petal_ratio"] = (
    df["petal_length"] /
    df["petal_width"]
)

df.groupby("species") ["petal_ratio"].mean()
```
