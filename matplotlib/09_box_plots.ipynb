# Box Plots
A box plot is a statistical visualization that summarizes the distribution of a dataset using five key numbers:
- Minimun (lowest non-outlier)
- First Quartile(Q1) - 25th percentile
- Median(Q2) - 50th percentile
- Third Quartile(Q3) - 75th percentile
- Maximum (highest non-outlier)

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [-5, 7, 8, 5, 6, 9, 4, 10, 12, 15, 23]

plt.boxplot(data)
plt.grid(True)
```

# Operations
- Multiple Datasets
- Horizontal Box Plot
- Show Mean
- Change Whisker Length

```python
# Multiple Datasets
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels = ["group1", "group2"])

plt.grid(True)
plt.show()
```

```python
# Horizontal Box Plot
plt.boxplot(group1, vert = False, showmeans = True) # Show Mean
plt.grid(True)
```

```python
# Change Whisker Length
data = [7, 8, 5, 6, 9, 4, 10, 12, 15, 23]

plt.boxplot(data, vert = False, showmeans = True, whis = 2.0)
plt.grid(True)
```
