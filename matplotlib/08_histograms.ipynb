# Histograms:
A  histogram shows the distribution of numberical data by:
- Dividing values into bins (ranges)
- Counting how many values fall into each bin (frequency)

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(0)

scores = np.random.normal(70, 10, 100) # normal distribution

plt.hist(scores, color="skyblue", edgecolor="black")
plt.show()
```

```python
plt.hist(scores, color="skyblue", edgecolor="black", bins=5)
plt.title("Distribution of marks for Students")
plt.ylabel("Number of Students")
plt.xlabel("Score")
plt.show()
```

# axvline:
- axvline is used to draw a vertical reference line on a plot, common to mark cutoffs, averages, pass marks, or thresholds.

```python
plt.hist(scores, color = "skyblue", edgecolor="black", bins=5)
plt.axvline(
    x = 85,
    label = "First Class (Above 85)",
    color = "red",
    linewidth = "2",
    linestyle = "--"
)

plt.title("Distribution of marks for Students")
plt.ylabel("Number of Students")
plt.xlabel("Score")
plt.show()
```
