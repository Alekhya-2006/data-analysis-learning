# Pie Charts:
A pie chart is a circular chart divided into slices to show relative propotions of a whole.
- Each slice = a category
- Size of slice = value of thet catogory relative to total

```python
import matplotlib.pyplot as plt

expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]


plt.pie(amounts, labels = expenses)
plt.style.use("classic")
plt.show()
```

# Customizations:
- autopct="%1.1f%%" --> show percentage of each slice
- colors --> add custom colours
- Wedgeprops --> define (edgecolor, linewidth, linestyle, alpha, fill...)
- explode --> highlights a slice by pulling it out
- startangle --> rotate the chart
- shadow --> adds 3D effect or makes the chart circular

```python
plt.pie(amounts, labels = expenses, autopct = "%1.1f%%",
        wedgeprops = {
            "edgecolor": "black",
            "linewidth": 1,
            "linestyle": "--"
        }
       )
plt.show()
```

```python
plt.pie(amounts, labels = expenses, autopct = "%1.1f%%",
        wedgeprops = {
            "fill": False
        })
plt.show()
```

```python
explode = [0, 0, 0, 0, 0.3]
plt.pie(amounts, labels = expenses, autopct = "%1.1f%%", explode = explode)
plt.show()
```

```python
plt.pie(amounts,
        labels = expenses, 
        autopct = "%1.1f%%",
        explode = explode,
        startangle = 180
       )
plt.show()
```

```python
plt.pie(amounts, labels = expenses,
        autopct = "%1.1f%%",
        explode = explode,
        shadow = True
       )
plt.show()
```
