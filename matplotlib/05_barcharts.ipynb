# Bar Charts:
A bar plot (or bar charts) is a graph that displays categorical data using rectangular bars.
Each bar's height(or length, in horizontal bars) represents a value.

```python
import matplotlib.pyplot as plt

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.bar(years, oscar_revenue, color = "grey")
plt.title("Revenue in each Year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+30, str(oscar_revenue[i]), ha = "center")

plt.ylim(0, max(oscar_revenue) + 100) #limit change
```

```python
import numpy as np

# Compare 2 sets of data

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar", 
    "Inception",
    "Hugo", 
    "Lincoln"
]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))

width = 0.4
plt.bar(x-width/2, oscar_revenue, width, label = "Oscar Movies") # shifting to LEFT
plt.bar(x+width/2, non_oscar_revenue, width, label = "Non-Oscar Movies") # shifting to RIGHT

plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.xticks(x, years)
```

```python
# Horizontal Charts
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.barh(oscar_movies, oscar_revenue)
plt.xlabel("Revenue (in $M)")
plt.ylabel("Movie Names")
```
