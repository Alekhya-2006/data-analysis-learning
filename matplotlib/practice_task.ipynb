```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Task

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
cities = ["New York", "London", "Delhi", "Tokyo"]
temperatures = [
    [22, 23, 21, 24, 25], # New York
    [18, 19, 17, 20, 21], # London
    [30, 32, 31, 33, 34], # Delhi
    [25, 26, 24, 27, 28]  # Tokyo
]
```

```python
fig, ax = plt.subplots(2,2)
city_no = 0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days, temperatures[city_no], marker = "o")
        ax[i][j].set_title(cities[city_no])
        ax[i][j].grid(True)
        ax[i][j].set_xlabel("Days")
        ax[i][j].set_ylabel("Temperature")
        city_no = city_no + 1

fig.suptitle("Temperatures in Cities over the week")
fig.tight_layout()   
```

```python
fig, axes = plt.subplots(2,2)

axes = axes.flatten()

for i, axes in enumerate(axes):
    axes.plot(days, temperatures[i], marker = "o")
    axes.set_title(cities[i])
    axes.grid(True)
    
fig.suptitle("Temperatures in Cities over the week")
fig.supylabel("Temperature")
fig.supxlabel("Days")

fig.tight_layout() 
```
