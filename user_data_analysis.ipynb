```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data    
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# Clean and Structure the data
def clean_data(data):
    text_to_num = {"one": 1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        
        # Clean ratings - data consistency
        rating = user["rating"]
        if isinstance(rating, str):
            raw_rating = rating.strip().lower()

            if raw_rating in text_to_num:
                user["rating"] = text_to_num[raw_rating]

        # Handle missing values
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = "Unknown"
            
        # Deduplication:
        clean_name = user["name"].strip()

        if clean_name in unique_users:
            continue

        unique_users.add(clean_name)  
        cleaned_data.append(user)    
        
    return cleaned_data    
```

```python
data = clean_data(data)
```

```python
# Get meaningful insights from data
def get_insights(data):
    
    # Avg rating
    tot_rating = 0
    for user in data:
        tot_rating += float(user["rating"])
    print(f'Average rating = {tot_rating / len(data)}')    

    # Percentage of users with poor rating
    poor_ratings = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_ratings += 1

    percentage = (poor_ratings / len(data)) * 100    
    print(f'Percentage of users with poor rating = {percentage}%')
```

```python
get_insights(data)
```

```python
# Build recommendation Feature
def get_recommendation(data):
    recommendations = []

    for user in data:
        current_recomm = {
            "name": user["name"]
        }
        
        if float(user["rating"]) >= 4:
            current_recomm["brand"] = "Apple"
        else:
            current_recomm["brand"] = "Samsung"

        recommendations.append(current_recomm)
    return recommendations
        
```

```python
get_recommendation(data)
```
